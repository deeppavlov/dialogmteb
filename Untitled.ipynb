```python
from mteb import MTEB
import torch
from sentence_transformers import SentenceTransformer
from mteb.tasks.Retrieval.rus.RuStatcanDialogueDatasetRetrieval import RuStatcanDialogueDatasetRetrieval
from mteb.tasks.Retrieval.rus.RuFaithDialRetrieval import RuFaithDialRetrieval
from mteb.tasks.Reranking.rus.RuWebLINXCandidatesReranking import RuWebLINXCandidatesReranking
from mteb.tasks.Reranking import WebLINXCandidatesReranking
from mteb.tasks.Retrieval import FaithDialRetrieval, StatcanDialogueDatasetRetrieval
from mteb.tasks.Classification.rus import RuAirDialogueClassification
from mteb.tasks.Classification.eng import AirDialogueClassification
import mteb
device = "cuda" if torch.cuda.is_available() else "cpu"

model = mteb.get_model("minishlab/potion-base-2M")

russian_datasets = [
    # RuFaithDialRetrieval(),
    # RuWebLINXCandidatesReranking(),
    AirDialogueClassification(),
    RuAirDialogueClassification()
]

for dataset_name in russian_datasets:
    print(f"\n{'='*20} Evaluating {dataset_name} {'='*20}")
    evaluation = MTEB(tasks=[dataset_name])
    
    results = evaluation.run(model, output_folder=f"results")
    
    print(f"Results for {dataset_name}:")
    if dataset_name in results:
        print(results[dataset_name])
    else:
        print("No results found for this dataset")

print("\nAll evaluations completed!")

try:
    print("\n\nRunning all Russian datasets together:")
    evaluation_all = MTEB(tasks=russian_datasets)
    results_all = evaluation_all.run(model, output_folder=f"results/1_all")
    print("Combined results:", results_all[0].scores)
except Exception as e:
    print(f"Error running combined evaluation: {e}")
```
