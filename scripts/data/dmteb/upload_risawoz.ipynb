```python
def split_dataset_by_domains(ds, mapping):
    """Split dataset by domains and filter columns to match each domain.

    Args:
        ds: HuggingFace dataset
        mapping: dict mapping domain names to lists of domain values

    Returns:
        dict: Dictionary with domain names as keys and filtered datasets as values
    """
    # Create reverse mapping from domain values to domain names
    domain_value_to_name = {}
    for domain_name, domain_values in mapping.items():
        for domain_value in domain_values:
            domain_value_to_name[domain_value] = domain_name

    def get_domain_from_column(column_name):
        """Extract domain from column name like 'inform-Hospital-metro station'"""
        if column_name.startswith("inform-"):
            parts = column_name.split("-")
            if len(parts) >= 2:
                potential_domain = parts[1]
                return domain_value_to_name.get(potential_domain)
        return None

    def filter_rows_and_columns_for_domain(examples, target_domain_name):
        """Filter rows that contain the target domain and keep only relevant columns"""
        target_domain_values = mapping[target_domain_name]

        # Get all column names for this domain
        relevant_columns = [
            "turn_id",
            "domains",
            "turn_domain",
            "text",
            "history",
            "dialogue_id",
        ]  # Always keep these
        for col in examples.keys():
            col_domain = get_domain_from_column(col)
            if col_domain == target_domain_name:
                relevant_columns.append(col)

        # Filter rows that contain any of the target domain values
        filtered_examples = {col: [] for col in relevant_columns}

        for i in range(len(examples["domains"])):
            # Check if any target domain value is in this row's domains
            row_domains = examples["domains"][i]
            if any(domain_val in row_domains for domain_val in target_domain_values):
                for col in relevant_columns:
                    filtered_examples[col].append(examples[col][i])

        return filtered_examples

    # Split dataset for each domain
    domain_datasets = {}

    for domain_name in mapping.keys():
        print(f"Processing domain: {domain_name}")

        # Filter dataset for this domain
        filtered_ds = {}
        for split_name in ds.keys():
            filtered_split = ds[split_name].map(
                lambda examples: filter_rows_and_columns_for_domain(
                    examples, domain_name
                ),
                batched=True,
                remove_columns=list(ds[split_name].features.keys()),
            )
            # Remove empty datasets
            if len(filtered_split) > 0:
                filtered_ds[split_name] = filtered_split

        if filtered_ds:  # Only add if we have data
            domain_datasets[domain_name] = filtered_ds
            print(f"  - Created dataset with {len(filtered_ds)} splits")
        else:
            print(f"  - No data found for domain {domain_name}")

    return domain_datasets
```

```python
import json
from collections import defaultdict

import pandas as pd

from datasets import load_dataset

# Load datasets for different languages
languages = ["en", "enhi", "fr", "hi", "ko", "zh"]
datasets = {}

print("Loading datasets for different languages...")
for lang in languages:
    try:
        datasets[lang] = load_dataset("DeepPavlov/XRISAWOZ", lang)
        print(
            f"✅ Loaded {lang}: {sum(len(ds) for ds in datasets[lang].values())} total rows"
        )
    except Exception as e:
        print(f"❌ Failed to load {lang}: {e}")


# Calculate statistics across languages
def calculate_cross_language_statistics(datasets, mapping):
    """Calculate statistics for each domain across all languages"""
    all_stats = defaultdict(
        lambda: defaultdict(lambda: defaultdict(lambda: defaultdict(dict)))
    )

    for lang, ds in datasets.items():
        print(f"\nProcessing {lang.upper()} language...")

        for split in ds.keys():
            for mapping_key, mapping_values in mapping.items():
                # Filter dataset for current domain
                filtered_ds = ds[split].filter(
                    lambda x: any([dom in mapping_values for dom in x["domains"]])
                )

                if len(filtered_ds) == 0:
                    continue

                print(f"  {split} - {mapping_key}: {len(filtered_ds)} rows")

                for col_name in filtered_ds.column_names:
                    if col_name.startswith("inform-"):
                        values = pd.value_counts(filtered_ds[col_name], normalize=True)

                        # Handle list columns
                        if len(values) > 0 and isinstance(values.index[0], list):
                            values.index = [str(x) for x in values.index]

                        # Store statistics
                        all_stats[lang][split][mapping_key][col_name] = values.to_dict()

    return all_stats


# Calculate statistics
print("Calculating cross-language statistics...")
cross_lang_stats = calculate_cross_language_statistics(datasets, mapping)

# Save statistics to file
with open("cross_language_stats.json", "w", encoding="utf-8") as f:
    json.dump(cross_lang_stats, f, ensure_ascii=False, indent=2)

print("Statistics saved to cross_language_stats.json")
```

```python
def create_cross_language_value_mappings(cross_lang_stats):
    """Create value mappings from all languages to English based on statistics"""

    def normalize_value(value):
        """Normalize value for comparison"""
        if not isinstance(value, str):
            return str(value).lower()

        # Basic normalization
        normalized = value.lower().strip()
        normalized = re.sub(
            r"[^\w\s-]", "", normalized
        )  # Remove special chars except dash
        normalized = re.sub(r"\s+", " ", normalized)  # Normalize spaces
        return normalized

    def find_english_equivalent(value, english_values):
        """Find the most likely English equivalent for a non-English value"""
        normalized_value = normalize_value(value)

        # Direct match
        if normalized_value in [normalize_value(en_val) for en_val in english_values]:
            for en_val in english_values:
                if normalize_value(en_val) == normalized_value:
                    return en_val

        # Fuzzy matching based on common patterns
        for en_val in english_values:
            en_normalized = normalize_value(en_val)

            # Check if they share significant word parts
            value_words = set(normalized_value.split())
            en_words = set(en_normalized.split())

            if value_words & en_words:  # If they have common words
                return en_val

            # Check substring matching
            if (len(normalized_value) > 3 and normalized_value in en_normalized) or (
                len(en_normalized) > 3 and en_normalized in normalized_value
            ):
                return en_val

        return None

    # Get English statistics as reference
    english_stats = cross_lang_stats.get("en", {})

    # Create mappings for each language to English
    language_mappings = defaultdict(lambda: defaultdict(dict))

    for lang, lang_stats in cross_lang_stats.items():
        if lang == "en":  # Skip English as it's our target
            continue

        print(f"\nCreating mappings for {lang.upper()} → English...")

        for split in lang_stats:
            for domain in lang_stats[split]:
                for column in lang_stats[split][domain]:
                    # Get values for this language
                    lang_values = set(lang_stats[split][domain][column].keys())

                    # Get corresponding English values
                    english_values = set()
                    if (
                        split in english_stats
                        and domain in english_stats[split]
                        and column in english_stats[split][domain]
                    ):
                        english_values = set(
                            english_stats[split][domain][column].keys()
                        )

                    if not english_values:
                        continue

                    # Create mappings
                    column_mappings = {}
                    for lang_value in lang_values:
                        if lang_value in ["none", "None", ""]:
                            continue

                        # Skip if it's already in English
                        if lang_value in english_values:
                            continue

                        english_equiv = find_english_equivalent(
                            lang_value, english_values
                        )
                        if english_equiv:
                            column_mappings[lang_value] = english_equiv
                            print(f"  {column}: '{lang_value}' → '{english_equiv}'")

                    if column_mappings:
                        language_mappings[lang][column].update(column_mappings)

    return language_mappings


# Create cross-language mappings
print("Creating cross-language value mappings...")
cross_lang_mappings = create_cross_language_value_mappings(cross_lang_stats)

# Save mappings
with open("cross_language_mappings.json", "w", encoding="utf-8") as f:
    json.dump(cross_lang_mappings, f, ensure_ascii=False, indent=2)

print("Cross-language mappings saved to cross_language_mappings.json")
```

```python
def create_value_tracking_system():
    """Create a system to track original values and their processed versions"""
    original_to_processed = defaultdict(dict)
    processed_to_original = defaultdict(dict)

    def track_mapping(column, original_value, processed_value):
        """Track a mapping between original and processed values"""
        if original_value != processed_value:
            original_to_processed[column][original_value] = processed_value
            processed_to_original[column][processed_value] = original_value

    return original_to_processed, processed_to_original, track_mapping


def enhanced_process_with_tracking(
    row, language_mappings, source_language, track_mapping
):
    """Process row while tracking all value changes - COMPLETE VERSION"""
    processed_row = row.copy()

    # Step 1: Apply cross-language mappings
    if source_language in language_mappings:
        lang_mappings = language_mappings[source_language]
        for column_name, value_mapping in lang_mappings.items():
            if (
                column_name in processed_row
                and processed_row[column_name] in value_mapping
            ):
                original_value = processed_row[column_name]
                new_value = value_mapping[original_value]
                processed_row[column_name] = new_value
                track_mapping(column_name, original_value, new_value)

    # Step 2: Apply ALL your original process_row transformations with tracking

    # PC-CPU transformations
    if processed_row.get("inform-PC-CPU") == "i7":
        track_mapping("inform-PC-CPU", "i7", "Intel Core i7")
        processed_row["inform-PC-CPU"] = "Intel Core i7"

    # Movie type transformations
    if processed_row.get("inform-Movie-type") in ("romantic movie", "romance"):
        original = processed_row["inform-Movie-type"]
        track_mapping("inform-Movie-type", original, "romantic")
        processed_row["inform-Movie-type"] = "romantic"
    if processed_row.get("inform-Movie-type") in (
        "action movie",
        "action movies",
        "action film",
    ):
        original = processed_row["inform-Movie-type"]
        track_mapping("inform-Movie-type", original, "action")
        processed_row["inform-Movie-type"] = "action"
    if processed_row.get("inform-Movie-type") in ("horror movies", "horror movie"):
        original = processed_row["inform-Movie-type"]
        track_mapping("inform-Movie-type", original, "horror")
        processed_row["inform-Movie-type"] = "horror"
    if processed_row.get("inform-Movie-type") in ("science fiction movie"):
        original = processed_row["inform-Movie-type"]
        track_mapping("inform-Movie-type", original, "science fiction")
        processed_row["inform-Movie-type"] = "science fiction"

    # Train classification transformations
    if processed_row.get("inform-Train-classification") in (
        "high-speed",
        "high-speed train",
    ):
        original = processed_row["inform-Train-classification"]
        track_mapping("inform-Train-classification", original, "high-speed train")
        processed_row["inform-Train-classification"] = "high-speed train"

    # Hotel room type transformations
    if processed_row.get("inform-Hotel-room type") in ("standard rooms"):
        original = processed_row["inform-Hotel-room type"]
        track_mapping("inform-Hotel-room type", original, "standard room")
        processed_row["inform-Hotel-room type"] = "standard room"
    if processed_row.get("inform-Hotel-room type") in ("king-size rooms"):
        original = processed_row["inform-Hotel-room type"]
        track_mapping("inform-Hotel-room type", original, "king-size room")
        processed_row["inform-Hotel-room type"] = "king-size room"

    # Attraction consumption transformations
    if processed_row.get("inform-Attraction-consumption") in (
        "a slightly more expensive"
    ):
        original = processed_row["inform-Attraction-consumption"]
        track_mapping(
            "inform-Attraction-consumption", original, "slightly more expensive"
        )
        processed_row["inform-Attraction-consumption"] = "slightly more expensive"

    # Remove attract score (track removal)
    if "inform-Attraction-score" in processed_row:
        track_mapping(
            "inform-Attraction-score",
            processed_row["inform-Attraction-score"],
            "[REMOVED]",
        )
        processed_row.pop("inform-Attraction-score", None)

    # Flight date transformations
    if processed_row.get("inform-Flight-date") in ("Next Thursday"):
        original = processed_row["inform-Flight-date"]
        track_mapping("inform-Flight-date", original, "next Thursday")
        processed_row["inform-Flight-date"] = "next Thursday"

    # PC usage transformations
    if processed_row.get("inform-PC-usage") in ("lightweight"):
        original = processed_row["inform-PC-usage"]
        track_mapping("inform-PC-usage", original, "lightweight laptop")
        processed_row["inform-PC-usage"] = "lightweight laptop"

    # Restaurant price range transformations
    if processed_row.get("inform-Restaurant-pricerange") in (
        "a slightly more expensive"
    ):
        original = processed_row["inform-Restaurant-pricerange"]
        track_mapping(
            "inform-Restaurant-pricerange", original, "slightly more expensive"
        )
        processed_row["inform-Restaurant-pricerange"] = "slightly more expensive"

    # Train seat type transformations
    if processed_row.get("inform-Train-seat type") in (
        "A first class",
        "a first-class ticket",
        "first-class seat",
    ):
        original = processed_row["inform-Train-seat type"]
        track_mapping("inform-Train-seat type", original, "first-class")
        processed_row["inform-Train-seat type"] = "first-class"
    if processed_row.get("inform-Train-seat type") in (
        "second class ticket",
        "second-class seat",
        "second-class ticket",
    ):
        original = processed_row["inform-Train-seat type"]
        track_mapping("inform-Train-seat type", original, "second class")
        processed_row["inform-Train-seat type"] = "second class"

    # Movie decade transformations
    if processed_row.get("inform-Movie-decade") in ("the 2000s"):
        original = processed_row["inform-Movie-decade"]
        track_mapping("inform-Movie-decade", original, "2000s")
        processed_row["inform-Movie-decade"] = "2000s"
    if processed_row.get("inform-Movie-decade") in ("the 1980s"):
        original = processed_row["inform-Movie-decade"]
        track_mapping("inform-Movie-decade", original, "1980s")
        processed_row["inform-Movie-decade"] = "1980s"

    # Weather date transformations
    if processed_row.get("inform-Weather-date") in ("Next Thursday"):
        original = processed_row["inform-Weather-date"]
        track_mapping("inform-Weather-date", original, "next Thursday")
        processed_row["inform-Weather-date"] = "next Thursday"

    # Hotel price range corrections
    if processed_row.get("inform-Hotel-pricerange") in ("Wuzhong District"):
        original = processed_row["inform-Hotel-pricerange"]
        track_mapping("inform-Hotel-pricerange", original, "none")
        processed_row["inform-Hotel-pricerange"] = "none"

    # Attraction type transformations
    if processed_row.get("inform-Attraction-type") in ("landscape scenic spots"):
        original = processed_row["inform-Attraction-type"]
        track_mapping("inform-Attraction-type", original, "landscape scenic spot")
        processed_row["inform-Attraction-type"] = "landscape scenic spot"
    if processed_row.get("inform-Attraction-type") in ("water towns"):
        original = processed_row["inform-Attraction-type"]
        track_mapping("inform-Attraction-type", original, "water town")
        processed_row["inform-Attraction-type"] = "water town"
    if processed_row.get("inform-Attraction-type") in (
        "historic site in Suzhou",
        "historical sites",
    ):
        original = processed_row["inform-Attraction-type"]
        track_mapping("inform-Attraction-type", original, "historical site")
        processed_row["inform-Attraction-type"] = "historical site"
    if processed_row.get("inform-Attraction-type") in ("Suzhou-styled garden"):
        original = processed_row["inform-Attraction-type"]
        track_mapping("inform-Attraction-type", original, "Suzhou garden")
        processed_row["inform-Attraction-type"] = "Suzhou garden"

    # Hospital departments transformations
    if processed_row.get("inform-Hospital-key departments") in ("pediatric clinic"):
        original = processed_row["inform-Hospital-key departments"]
        track_mapping("inform-Hospital-key departments", original, "pediatrics")
        processed_row["inform-Hospital-key departments"] = "pediatrics"
    if processed_row.get("inform-Hospital-key departments") in (
        "Gynecology Department"
    ):
        original = processed_row["inform-Hospital-key departments"]
        track_mapping(
            "inform-Hospital-key departments", original, "gynecology department"
        )
        processed_row["inform-Hospital-key departments"] = "gynecology department"

    # Remove car power level (track removal)
    if "inform-Car-power level" in processed_row:
        track_mapping(
            "inform-Car-power level",
            processed_row["inform-Car-power level"],
            "[REMOVED]",
        )
        processed_row.pop("inform-Car-power level", None)

    # PC price range transformations
    if processed_row.get("inform-PC-pricerange") in (
        "5,000-10,000 yuan",
        "5,000 to 10,000",
    ):
        original = processed_row["inform-PC-pricerange"]
        track_mapping("inform-PC-pricerange", original, "5,000-10,000")
        processed_row["inform-PC-pricerange"] = "5,000-10,000"
    if processed_row.get("inform-PC-pricerange") in (
        "below 5,000",
        "less than 5,000 yuan",
        "No more than 5,000 yuan",
        "within 5,000 yuan",
    ):
        original = processed_row["inform-PC-pricerange"]
        track_mapping("inform-PC-pricerange", original, "less than 5,000")
        processed_row["inform-PC-pricerange"] = "less than 5,000"

    # Restaurant cuisine transformations
    if processed_row.get("inform-Restaurant-cuisine") in (
        "Jiangsu Zhejiang Cuisine",
        "Jiangsu and Zhejiang cuisine",
        "Jiangsu Zhejiang dishes",
    ):
        original = processed_row["inform-Restaurant-cuisine"]
        track_mapping("inform-Restaurant-cuisine", original, "Jiangsu Zhejiang cuisine")
        processed_row["inform-Restaurant-cuisine"] = "Jiangsu Zhejiang cuisine"
    if processed_row.get("inform-Restaurant-cuisine") in ("HotPot", "hot pot"):
        original = processed_row["inform-Restaurant-cuisine"]
        track_mapping("inform-Restaurant-cuisine", original, "hot pot")
        processed_row["inform-Restaurant-cuisine"] = "hot pot"
    if processed_row.get("inform-Restaurant-cuisine") in (
        "Japanese restaurant",
        "Japanese food",
        "Japanese",
    ):
        original = processed_row["inform-Restaurant-cuisine"]
        track_mapping("inform-Restaurant-cuisine", original, "Japanese")
        processed_row["inform-Restaurant-cuisine"] = "Japanese"
    if processed_row.get("inform-Restaurant-cuisine") in (
        "bread and dessert",
        "bread and desserts",
        "bread",
    ):
        original = processed_row["inform-Restaurant-cuisine"]
        track_mapping("inform-Restaurant-cuisine", original, "bread and dessert")
        processed_row["inform-Restaurant-cuisine"] = "bread and dessert"
    if processed_row.get("inform-Restaurant-cuisine") in (
        "Korean foods",
        "Korean food",
        "Korean Cuisine",
        "South Korean cuisine",
    ):
        original = processed_row["inform-Restaurant-cuisine"]
        track_mapping("inform-Restaurant-cuisine", original, "Korean")
        processed_row["inform-Restaurant-cuisine"] = "Korean"
    if processed_row.get("inform-Restaurant-cuisine") in (
        "Northwestern",
        "Northwest cuisine",
    ):
        original = processed_row["inform-Restaurant-cuisine"]
        track_mapping("inform-Restaurant-cuisine", original, "Northwest")
        processed_row["inform-Restaurant-cuisine"] = "Northwest"
    if processed_row.get("inform-Restaurant-cuisine") in (
        "fast food",
        "fast food and quick meals",
    ):
        original = processed_row["inform-Restaurant-cuisine"]
        track_mapping("inform-Restaurant-cuisine", original, "fast food")
        processed_row["inform-Restaurant-cuisine"] = "fast food"

    # TV production country transformations
    if processed_row.get("inform-TV-production country or area") in (
        "China mainland",
        "Chinese mainland TV show",
    ):
        original = processed_row["inform-TV-production country or area"]
        track_mapping(
            "inform-TV-production country or area", original, "Chinese mainland TV show"
        )
        processed_row["inform-TV-production country or area"] = (
            "Chinese mainland TV show"
        )
    if processed_row.get("inform-TV-production country or area") in (
        "Korean",
        "Korean TV show",
        "Korean series",
        "Korean drama",
    ):
        original = processed_row["inform-TV-production country or area"]
        track_mapping(
            "inform-TV-production country or area", original, "Korean TV show"
        )
        processed_row["inform-TV-production country or area"] = "Korean TV show"
    if processed_row.get("inform-TV-production country or area") in (
        "Taiwan, China",
        "a TV show from Taiwan, China",
    ):
        original = processed_row["inform-TV-production country or area"]
        track_mapping("inform-TV-production country or area", original, "Taiwan, China")
        processed_row["inform-TV-production country or area"] = "Taiwan, China"
    if processed_row.get("inform-TV-production country or area") in (
        "Japanese TV show",
        "Japanese",
        "Japanese shows",
    ):
        original = processed_row["inform-TV-production country or area"]
        track_mapping(
            "inform-TV-production country or area", original, "Japanese TV show"
        )
        processed_row["inform-TV-production country or area"] = "Japanese TV show"
    if processed_row.get("inform-TV-production country or area") in (
        "American",
        "American TV show",
        "American series",
        "in America",
    ):
        original = processed_row["inform-TV-production country or area"]
        track_mapping(
            "inform-TV-production country or area", original, "American TV show"
        )
        processed_row["inform-TV-production country or area"] = "American TV show"

    # Remove PC CPU model (track removal)
    if "inform-PC-CPU model" in processed_row:
        track_mapping(
            "inform-PC-CPU model", processed_row["inform-PC-CPU model"], "[REMOVED]"
        )
        processed_row.pop("inform-PC-CPU model", None)

    # Flight class cabin transformations
    if processed_row.get("inform-Flight-class cabin") in ("economy class", "economy"):
        original = processed_row["inform-Flight-class cabin"]
        track_mapping("inform-Flight-class cabin", original, "economy class")
        processed_row["inform-Flight-class cabin"] = "economy class"
    if processed_row.get("inform-Flight-class cabin") in (
        "Business class",
        "A business class",
        "first class",
    ):
        original = processed_row["inform-Flight-class cabin"]
        track_mapping("inform-Flight-class cabin", original, "business class")
        processed_row["inform-Flight-class cabin"] = "business class"

    # Remove flight arrival time (track removal)
    if "inform-Flight-arrival time" in processed_row:
        track_mapping(
            "inform-Flight-arrival time",
            processed_row["inform-Flight-arrival time"],
            "[REMOVED]",
        )
        processed_row.pop("inform-Flight-arrival time", None)

    # Movie production country transformations
    if processed_row.get("inform-Movie-production country or area") in (
        "China mainland",
        "Chinese mainland movie",
    ):
        original = processed_row["inform-Movie-production country or area"]
        track_mapping(
            "inform-Movie-production country or area",
            original,
            "Chinese mainland movie",
        )
        processed_row["inform-Movie-production country or area"] = (
            "Chinese mainland movie"
        )
    if processed_row.get("inform-Movie-production country or area") in (
        "Taiwan, China",
        "Taiwan",
    ):
        original = processed_row["inform-Movie-production country or area"]
        track_mapping(
            "inform-Movie-production country or area", original, "Taiwan, China"
        )
        processed_row["inform-Movie-production country or area"] = "Taiwan, China"
    if processed_row.get("inform-Movie-production country or area") in (
        "Hong Kong",
        "Hong Kong TV show",
    ):
        original = processed_row["inform-Movie-production country or area"]
        track_mapping("inform-Movie-production country or area", original, "Hong Kong")
        processed_row["inform-Movie-production country or area"] = "Hong Kong"
    if processed_row.get("inform-Movie-production country or area") in (
        "Korea",
        "Korean",
        "Korean",
    ):
        original = processed_row["inform-Movie-production country or area"]
        track_mapping("inform-Movie-production country or area", original, "Korea")
        processed_row["inform-Movie-production country or area"] = "Korea"
    if processed_row.get("inform-Movie-production country or area") in (
        "in America",
        "the United States",
        "America",
        "in America",
    ):
        original = processed_row["inform-Movie-production country or area"]
        track_mapping(
            "inform-Movie-production country or area", original, "American TV show"
        )
        processed_row["inform-Movie-production country or area"] = "American TV show"

    # TV type transformations
    if processed_row.get("inform-TV-type") in (
        "romantic",
        "romantic TV show",
        "romance",
    ):
        original = processed_row["inform-TV-type"]
        track_mapping("inform-TV-type", original, "romantic")
        processed_row["inform-TV-type"] = "romantic"
    if processed_row.get("inform-TV-type") in (
        "feature TV show",
        "feature",
        "feature shows",
    ):
        original = processed_row["inform-TV-type"]
        track_mapping("inform-TV-type", original, "feature")
        processed_row["inform-TV-type"] = "feature"
    if processed_row.get("inform-TV-type") in (
        "crime",
        "crime drama",
        "crime TV show",
        "crime shows",
    ):
        original = processed_row["inform-TV-type"]
        track_mapping("inform-TV-type", original, "crime")
        processed_row["inform-TV-type"] = "crime"
    if processed_row.get("inform-TV-type") in (
        "action",
        "action TV show",
        "action shows",
    ):
        original = processed_row["inform-TV-type"]
        track_mapping("inform-TV-type", original, "action")
        processed_row["inform-TV-type"] = "action"
    if processed_row.get("inform-TV-type") in (
        "sci-fi show",
        "sci-fis",
        "scientific TV shows",
        "scientific",
        "science fiction",
    ):
        original = processed_row["inform-TV-type"]
        track_mapping("inform-TV-type", original, "science fiction")
        processed_row["inform-TV-type"] = "science fiction"

    # Remove restaurant per capita consumption (track removal)
    if "inform-Restaurant-per capita consumption" in processed_row:
        track_mapping(
            "inform-Restaurant-per capita consumption",
            processed_row["inform-Restaurant-per capita consumption"],
            "[REMOVED]",
        )
        processed_row.pop("inform-Restaurant-per capita consumption", None)

    # Attraction suitable people transformations
    if processed_row.get("inform-Attraction-the most suitable people") in (
        "friends",
        "friend",
        "go with friends",
        "besties",
        "travel with friends",
    ):
        original = processed_row["inform-Attraction-the most suitable people"]
        track_mapping("inform-Attraction-the most suitable people", original, "friends")
        processed_row["inform-Attraction-the most suitable people"] = "friends"
    if processed_row.get("inform-Attraction-the most suitable people") in (
        "go with kids",
        "kids",
        "family trip",
        "families",
        "family-friendly",
        "family",
        "traveling with family",
    ):
        original = processed_row["inform-Attraction-the most suitable people"]
        track_mapping(
            "inform-Attraction-the most suitable people", original, "family trip"
        )
        processed_row["inform-Attraction-the most suitable people"] = "family trip"
    if processed_row.get("inform-Attraction-the most suitable people") in (
        "girlfriend",
        "date",
    ):
        original = processed_row["inform-Attraction-the most suitable people"]
        track_mapping(
            "inform-Attraction-the most suitable people", original, "girlfriend"
        )
        processed_row["inform-Attraction-the most suitable people"] = "girlfriend"

    # Class time transformations
    if processed_row.get("inform-Class-time") in (
        "evening or mornings",
        "mornings or evening",
    ):
        original = processed_row["inform-Class-time"]
        track_mapping("inform-Class-time", original, "evening or morning")
        processed_row["inform-Class-time"] = "evening or morning"

    # Hotel star transformations
    if processed_row.get("inform-Hotel-star") in ("5", "5-star"):
        original = processed_row["inform-Hotel-star"]
        track_mapping("inform-Hotel-star", original, "5")
        processed_row["inform-Hotel-star"] = "5"
    if processed_row.get("inform-Hotel-star") in ("4", "4-star"):
        original = processed_row["inform-Hotel-star"]
        track_mapping("inform-Hotel-star", original, "4")
        processed_row["inform-Hotel-star"] = "4"
    if processed_row.get("inform-Hotel-star") in ("3", "3-star"):
        original = processed_row["inform-Hotel-star"]
        track_mapping("inform-Hotel-star", original, "3")
        processed_row["inform-Hotel-star"] = "3"
    if processed_row.get("inform-Hotel-star") in ("2", "2-star"):
        original = processed_row["inform-Hotel-star"]
        track_mapping("inform-Hotel-star", original, "2")
        processed_row["inform-Hotel-star"] = "2"
    if processed_row.get("inform-Hotel-star") in ("1", "1-star"):
        original = processed_row["inform-Hotel-star"]
        track_mapping("inform-Hotel-star", original, "1")
        processed_row["inform-Hotel-star"] = "1"

    return processed_row


def create_reverse_mapping_dataset(original_ds, processed_ds, processed_to_original):
    """Create a dataset that maps processed values back to their originals"""

    def add_original_values(row):
        """Add original values as new columns"""
        enhanced_row = row.copy()

        for column_name in row.keys():
            if (
                column_name.startswith("inform-")
                and column_name in processed_to_original
            ):
                processed_value = row[column_name]
                if processed_value in processed_to_original[column_name]:
                    original_value = processed_to_original[column_name][processed_value]
                    # Add original value as new column
                    # enhanced_row[f"{column_name}_original"] = original_value
                    enhanced_row[column_name] = original_value
                else:
                    # Value wasn't changed
                    # enhanced_row[f"{column_name}_original"] = processed_value
                    enhanced_row[column_name] = processed_value

        return enhanced_row

    return processed_ds.map(add_original_values)


def process_all_languages_with_tracking():
    """Process all languages while tracking value changes"""
    # Load cross-language mappings
    with open("cross_language_mappings.json", encoding="utf-8") as f:
        cross_lang_mappings = json.load(f)

    all_results = {}
    all_mappings_summary = defaultdict(lambda: defaultdict(int))

    for lang, ds in datasets.items():
        # if lang != "ko":
        #     continue
        # Remove the filter - process ALL languages
        print(f"\n{'=' * 50}")
        print(f"PROCESSING {lang.upper()} WITH VALUE TRACKING")
        print(f"{'=' * 50}")

        # Initialize tracking
        original_to_processed, processed_to_original, track_mapping = (
            create_value_tracking_system()
        )

        # Process with tracking
        ds_processed = ds.map(
            lambda row: enhanced_process_with_tracking(
                row, cross_lang_mappings, lang, track_mapping
            ),
            num_proc=4,
        )

        # Create dataset with original values preserved
        ds_with_originals = create_reverse_mapping_dataset(
            ds, ds_processed, processed_to_original
        )

        # Split by domains
        domain_datasets = split_dataset_by_domains(ds_with_originals, mapping)

        # Save mapping information
        lang_mapping_summary = {}
        for column, mappings in original_to_processed.items():
            lang_mapping_summary[column] = dict(mappings)
            all_mappings_summary[lang][column] = len(mappings)

        # Save detailed mappings for this language
        with open(f"value_mappings_{lang}.json", "w", encoding="utf-8") as f:
            json.dump(
                {
                    "original_to_processed": dict(original_to_processed),
                    "processed_to_original": dict(processed_to_original),
                },
                f,
                ensure_ascii=False,
                indent=2,
            )

        print(f"Saved detailed mappings to value_mappings_{lang}.json")

        # Upload domains
        lang_results = {}
        for domain, domain_ds in domain_datasets.items():
            domain_clean = domain.lower().replace(" ", "_")
            config_name = f"{lang}_{domain_clean}"

            try:
                ds_dict = DatasetDict(domain_ds)

                # Show sample with original values
                if len(list(domain_ds.values())[0]) > 0:
                    sample = list(domain_ds.values())[0][0]
                    original_cols = [
                        k for k in sample.keys() if k.endswith("_original")
                    ][:3]
                    if original_cols:
                        print("  📋 Sample original values:")
                        for col in original_cols:
                            base_col = col.replace("_original", "")
                            if base_col in sample and sample[base_col] != sample[col]:
                                print(
                                    f"    • {base_col}: '{sample[col]}' → '{sample[base_col]}'"
                                )
                ds_dict.push_to_hub(
                    "DeepPavlov/XRISAWOZ",
                    config_name,
                    commit_message=f"Add {lang} {domain} with original values preserved and complete processing",
                )

                total_rows = sum(len(split_ds) for split_ds in domain_ds.values())
                lang_results[domain] = f"✅ {total_rows:,} rows"
                print(f"  ✅ {domain}: {total_rows:,} rows uploaded")

            except Exception as e:
                lang_results[domain] = f"❌ {str(e)}"
                print(f"  ❌ {domain}: {e}")

        all_results[lang] = {
            "upload_results": lang_results,
            "mappings_count": sum(
                len(mappings) for mappings in original_to_processed.values()
            ),
        }

    return all_results, all_mappings_summary


# Run the processing with tracking
print("Starting processing with value tracking...")
results, mappings_summary = process_all_languages_with_tracking()

# Save overall summary
with open("all_language_mappings_summary.json", "w", encoding="utf-8") as f:
    json.dump(dict(mappings_summary), f, ensure_ascii=False, indent=2)

# Print comprehensive summary
print("\n" + "=" * 70)
print("🔄 VALUE TRACKING & PROCESSING SUMMARY")
print("=" * 70)

for lang, lang_data in results.items():
    print(f"\n📍 {lang.upper()} LANGUAGE:")
    print(f"  🔀 Value mappings created: {lang_data['mappings_count']:,}")

    upload_results = lang_data["upload_results"]
    success_count = sum(
        1 for result in upload_results.values() if result.startswith("✅")
    )
    print(f"  📤 Upload success: {success_count}/{len(upload_results)} domains")

    for domain, result in upload_results.items():
        print(f"    • {domain}: {result}")

print("\n📊 OVERALL STATISTICS:")
total_mappings = sum(lang_data["mappings_count"] for lang_data in results.values())
print(f"  • Total value transformations tracked: {total_mappings:,}")
print(f"  • Languages processed: {len(results)}")
print(f"  • Individual mapping files created: {len(results)} files")
print("  • Each dataset now includes both processed and original values")

print("\n💾 FILES CREATED:")
print("  • cross_language_stats.json - Raw statistics")
print("  • cross_language_mappings.json - Language mappings")
print("  • value_mappings_[lang].json - Detailed mappings per language")
print("  • all_language_mappings_summary.json - Summary of all mappings")
```
