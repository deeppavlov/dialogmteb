```python
from datasets import load_dataset
import pandas as pd
import json
```

```python
from collections import defaultdict


datasets_names = [("DeepPavlov/RISAWOZ", None), ("DeepPavlov/XRISAWOZ", "en"), ("DeepPavlov/Multi2WOZ", "ru"), ("DeepPavlov/MultiWOZ-2.1", None), ("DeepPavlov/BiToD", "en")]
for ds_names, config in datasets_names:
    ds = load_dataset(ds_names, config)
    ds_res = defaultdict(dict)
    for split in ds.keys():
        for col_name in ds["test"].column_names:
            values = pd.value_counts(ds["test"][col_name], normalize=True)
            if values.shape[0] > 30:
                continue
            if isinstance(values.index[0], list):
                values.index = [str(x) for x in values.index]
            ds_res[split][col_name] = values.to_dict()

    with open(ds_names.replace("/", "_") + ".json", "w") as f:
        json.dump(ds_res, f, indent=4, ensure_ascii=False)
```

```python
# Counter({('Flight', 'Weather'): 59,
#          ('Attraction', 'Restaurant', 'Hotel'): 48,
#          ('Attraction', 'Hotel'): 47,
#          ('PC',): 44,
#          ('Hotel',): 43,
#          ('Movie', 'TV'): 37,
#          ('Car',): 35,
#          ('Attraction',): 35,
#          ('Train',): 35,
#          ('Class',): 33,
#          ('Attraction', 'Restaurant'): 32,
#          ('Weather',): 32,
#          ('TV',): 30,
#          ('Train', 'Weather'): 29,
#          ('Restaurant',): 27,
#          ('Movie',): 26,
#          ('Flight',): 26,
#          ('Restaurant', 'Hotel'): 24,
#          ('Hospital',): 17})
mapping = {
    "Transport": ["Flight", "Weather",],
    "Train": ["Train"],
    "Car": ["Car"],
    "Attraction": ["Attraction", "Restaurant", "Hotel"],
    "PC": ["PC"],
    "Movie": ["Movie", "TV"],
    "Class": ["Class"],
    "Hospital": ["Hospital"],
}
```

```python
from collections import Counter, defaultdict
ds_res = defaultdict(lambda: defaultdict(dict))
for split in ds.keys():
    for mapping_key, mapping_values in mapping.items():
        filtered_ds = ds[split].filter(lambda x: any([dom in mapping_values for dom in x["domains"]]))
        for col_name in filtered_ds.column_names:
            values = pd.value_counts(filtered_ds[col_name], normalize=True)
            # skip text columns
            if values.shape[0] > 30:
                continue
            if isinstance(values.index[0], list):
                values.index = [str(x) for x in values.index]
            ds_res[split][mapping_key][col_name] = values.to_dict()

with open("DeepPavlov__XRISAWOZ" + ".json", "w") as f:
    json.dump(ds_res, f, indent=4, ensure_ascii=False)
```
