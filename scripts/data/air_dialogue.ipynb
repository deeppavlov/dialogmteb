```python
from __future__ import annotations

import json
from pathlib import Path

from tqdm import tqdm

import datasets
```

```python
folder = Path("/home/samoed/Downloads/airdialogue_data_1.3/airdialogue/")
```

```python
js = list(folder.glob("*_data.json"))
```

```python
js
```

```python
dataset = {}
for path in js[::-1]:
    split_name = path.name.split("_")[0]
    split = []
    with path.open() as f:
        for line in f:
            split.append(json.loads(line))

    parsed = []
    for row in tqdm(split):
        dialogue = row["dialogue"]
        parsed_dialogue = []
        for replic in dialogue:
            role = "user" if replic.startswith("customer") else "assistant"
            replic_content = replic.replace("agent: ", "").replace("customer: ", "")
            parsed_dialogue.append({"role": role, "content": replic_content})
        action = row["expected_action"]["status"]
        parsed.append(
            {
                "text": parsed_dialogue,
                "label": action,
                "action": row["action"],
                "intent": row["intent"],
                "expected_action": row["expected_action"],
            }
        )

    dataset[split_name] = parsed
```

```python
ds_dict = {k: datasets.Dataset.from_list(v) for k, v in dataset.items()}
```

```python
ds_dict = datasets.DatasetDict(ds_dict)
```
